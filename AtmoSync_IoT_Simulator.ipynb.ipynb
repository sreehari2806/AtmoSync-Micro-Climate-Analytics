```python
from datetime import datetime, timedelta

sensor_data = []

start_time = datetime.now()

for i in range(100):

    container_id = random.choice(container_ids)

    temperature = round(random.uniform(2, 15), 2)
    humidity = round(random.uniform(40, 90), 2)
    vibration = round(random.uniform(0.1, 2.0), 2)

    timestamp = start_time + timedelta(minutes=i)

    reading = {
        "timestamp": timestamp,
        "container_id": container_id,
        "commodity": commodities[container_id],
        "temperature": temperature,
        "humidity": humidity,
        "vibration": vibration
    }

    sensor_data.append(reading)

df = pd.DataFrame(sensor_data)

print("100 sensor readings generated successfully!")
```

```python
temperature_limits = {
    "Tomato": (2, 12),
    "Banana": (2, 14),
    "Mango": (2, 12),
    "Fish": (0, 4),
    "Milk": (2, 6)
}
```

```python
def check_temperature(row):

    minimum, maximum = temperature_limits[row["commodity"]]

    if row["temperature"] < minimum:
        return "Low Temperature"

    elif row["temperature"] > maximum:
        return "High Temperature"

    else:
        return "Normal"
```

```python
df["temperature_status"] = df.apply(check_temperature, axis=1)

df.head(10)
```

```python
df["temperature_status"].value_counts()
```

```python
df.to_csv("sensor_readings.csv", index=False)
```

```python
def check_humidity(row):

    if row["humidity"] < 45:
        return "Low Humidity"

    elif row["humidity"] > 85:
        return "High Humidity"

    else:
        return "Normal"
```

```python
df["humidity_status"] = df.apply(check_humidity, axis=1)

df.head(10)
```

```python
def check_container_status(row):

    if (row["temperature_status"] != "Normal" or
        row["humidity_status"] != "Normal"):

        return "Alert"

    else:
        return "Normal"
```

```python
df["container_status"] = df.apply(check_container_status, axis=1)

df.head(10)
```

```python
insert_sql = """
INSERT INTO ATMOSYNC_DB.RAW.SENSOR_READINGS_RAW
(
    TIMESTAMP,
    CONTAINER_ID,
    COMMODITY,
    TEMPERATURE,
    HUMIDITY,
    VIBRATION,
    TEMPERATURE_STATUS,
    HUMIDITY_STATUS,
    CONTAINER_STATUS
)
VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s)
"""

print("Snowflake insert query prepared!")
```

```python
from datetime import datetime

total_inserted = 0

while True:

    records = consumer.poll(
        timeout_ms=5000,
        max_records=100
    )

    if not records:
        print("No more messages available.")
        break

    rows = []

    for messages in records.values():
        for message in messages:

            data = message.value

            timestamp = data.get("timestamp")

            if timestamp:
                timestamp = datetime.fromisoformat(
                    timestamp.replace("Z", "+00:00")
                )

            row = (
                timestamp,
                data.get("container_id"),
                data.get("commodity"),
                data.get("temperature"),
                data.get("humidity"),
                data.get("vibration"),
                data.get("temperature_status"),
                data.get("humidity_status"),
                data.get("container_status")
            )

            rows.append(row)

    if rows:

        insert_cursor = conn.cursor()

        try:
            insert_cursor.executemany(insert_sql, rows)
            conn.commit()

            consumer.commit()

            total_inserted += len(rows)

            print("Inserted:", total_inserted)

        finally:
            insert_cursor.close()

print("Total new records inserted:", total_inserted)
```
