```python
temperature_limits = {
    "Tomato": (2, 12),
    "Banana": (2, 14),
    "Mango": (2, 12),
    "Fish": (0, 4),
    "Milk": (2, 6)
}
```

```python
def check_temperature(row):

    minimum, maximum = temperature_limits[row["commodity"]]

    if row["temperature"] < minimum:
        return "Low Temperature"

    elif row["temperature"] > maximum:
        return "High Temperature"

    else:
        return "Normal"
```

```python
df["temperature_status"] = df.apply(check_temperature, axis=1)

df.head(10)
```

```python
def check_humidity(row):

    if row["humidity"] < 45:
        return "Low Humidity"

    elif row["humidity"] > 85:
        return "High Humidity"

    else:
        return "Normal"
```

```python
from datetime import datetime

total_inserted = 0

while True:

    records = consumer.poll(
        timeout_ms=5000,
        max_records=100
    )

    if not records:
        print("No more messages available.")
        break

    rows = []

    for messages in records.values():
        for message in messages:

            data = message.value

            timestamp = data.get("timestamp")

            if timestamp:
                timestamp = datetime.fromisoformat(
                    timestamp.replace("Z", "+00:00")
                )

            row = (
                timestamp,
                data.get("container_id"),
                data.get("commodity"),
                data.get("temperature"),
                data.get("humidity"),
                data.get("vibration"),
                data.get("temperature_status"),
                data.get("humidity_status"),
                data.get("container_status")
            )

            rows.append(row)

    if rows:

        insert_cursor = conn.cursor()

        try:
            insert_cursor.executemany(insert_sql, rows)
            conn.commit()

            consumer.commit()

            total_inserted += len(rows)

            print("Inserted:", total_inserted)

        finally:
            insert_cursor.close()

print("Total new records inserted:", total_inserted)
```

```python
import json
from datetime import datetime

# Step A: Use the message already fetched in cell 67, if available.
# Otherwise, fetch a small batch from Kafka.

pending_messages = []

if "records" in globals() and records:
    for message_list in records.values():
        pending_messages.extend(message_list)

    records = {}

else:
    new_records = consumer.poll(
        timeout_ms=5000,
        max_records=5
    )

    for message_list in new_records.values():
        pending_messages.extend(message_list)

print("Messages received:", len(pending_messages))


# Step B: Validate and prepare the messages.

required_fields = [
    "timestamp",
    "container_id",
    "commodity",
    "temperature",
    "humidity",
    "vibration",
    "temperature_status",
    "humidity_status",
    "container_status"
]

valid_rows = []

for message in pending_messages:
    data = message.value

    if all(field in data and data[field] is not None
           for field in required_fields):

        timestamp = datetime.fromisoformat(
            str(data["timestamp"]).replace("Z", "+00:00")
        )

        valid_rows.append((
            timestamp,
            data["container_id"],
            data["commodity"],
            data["temperature"],
            data["humidity"],
            data["vibration"],
            data["temperature_status"],
            data["humidity_status"],
            data["container_status"]
        ))

    else:
        print("Skipping incomplete message:", data)


# Step C: Insert valid messages into staging and MERGE into raw.

if valid_rows:

    cursor = conn.cursor()

    try:
        cursor.execute("BEGIN")

        cursor.executemany("""
            INSERT INTO ATMOSYNC_DB.RAW.SENSOR_READINGS_STAGING (
                TIMESTAMP,
                CONTAINER_ID,
                COMMODITY,
                TEMPERATURE,
                HUMIDITY,
                VIBRATION,
                TEMPERATURE_STATUS,
                HUMIDITY_STATUS,
                CONTAINER_STATUS
            )
            VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s)
        """, valid_rows)

        cursor.execute("""
            MERGE INTO ATMOSYNC_DB.RAW.SENSOR_READINGS_RAW AS target

            USING (
                SELECT *
                FROM ATMOSYNC_DB.RAW.SENSOR_READINGS_STAGING

                QUALIFY ROW_NUMBER() OVER (
                    PARTITION BY TIMESTAMP, CONTAINER_ID
                    ORDER BY INGESTED_AT
                ) = 1
            ) AS source

            ON target.TIMESTAMP = source.TIMESTAMP
            AND target.CONTAINER_ID = source.CONTAINER_ID

            WHEN NOT MATCHED THEN
            INSERT (
                TIMESTAMP,
                CONTAINER_ID,
                COMMODITY,
                TEMPERATURE,
                HUMIDITY,
                VIBRATION,
                TEMPERATURE_STATUS,
                HUMIDITY_STATUS,
                CONTAINER_STATUS
            )
            VALUES (
                source.TIMESTAMP,
                source.CONTAINER_ID,
                source.COMMODITY,
                source.TEMPERATURE,
                source.HUMIDITY,
                source.VIBRATION,
                source.TEMPERATURE_STATUS,
                source.HUMIDITY_STATUS,
                source.CONTAINER_STATUS
            )
        """)

        cursor.execute("""
            DELETE FROM ATMOSYNC_DB.RAW.SENSOR_READINGS_STAGING
        """)

        conn.commit()

        # Commit Kafka offsets only after Snowflake succeeds.
        consumer.commit()

        print("Snowflake ingestion completed successfully!")
        print("Messages processed:", len(valid_rows))

    except Exception as error:
        conn.rollback()
        print("Ingestion failed:", error)
        print("Snowflake transaction rolled back.")

    finally:
        cursor.close()

else:
    print("No valid messages to insert. No offsets committed.")
```


## 10. Automated Kafka-to-Snowflake Data Ingestion

This section implements an automated data ingestion pipeline that transfers sensor readings from Apache Kafka into Snowflake.

The Python Kafka consumer retrieves incoming sensor messages, validates the records, and temporarily stores them in a Snowflake staging table.

A Snowflake MERGE operation then inserts only new records into the raw table, preventing duplicate records based on the timestamp and container ID.

After successful ingestion, the staging table is cleared and Kafka offsets are committed.


```python

from datetime import datetime

def ingest_kafka_to_snowflake():

    cursor = conn.cursor()

    try:
        # Step 1: Retrieve messages from Kafka
        records = consumer.poll(timeout_ms=10000, max_records=100)

        messages_received = []

        for partition_messages in records.values():
            for message in partition_messages:
                messages_received.append(message.value)

        print("Messages received:", len(messages_received))

        if not messages_received:
            print("No new messages available.")
            return

        # Step 2: Validate incoming records
        required_fields = [
            "timestamp", "container_id", "commodity",
            "temperature", "humidity", "vibration",
            "temperature_status", "humidity_status",
            "container_status"
        ]

        valid_records = []

        for data in messages_received:

            if not all(
                field in data and data[field] is not None
                for field in required_fields
            ):
                raise ValueError("Invalid sensor record detected.")

            valid_records.append((
                datetime.fromisoformat(str(data["timestamp"])),
                data["container_id"],
                data["commodity"],
                data["temperature"],
                data["humidity"],
                data["vibration"],
                data["temperature_status"],
                data["humidity_status"],
                data["container_status"]
            ))

        # Step 3: Insert records into staging
        cursor.executemany("""
            INSERT INTO ATMOSYNC_DB.RAW.SENSOR_READINGS_STAGING (
                TIMESTAMP, CONTAINER_ID, COMMODITY,
                TEMPERATURE, HUMIDITY, VIBRATION,
                TEMPERATURE_STATUS, HUMIDITY_STATUS,
                CONTAINER_STATUS
            )
            VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s)
        """, valid_records)

        # Step 4: Merge only new records into raw table
        cursor.execute("""
            MERGE INTO ATMOSYNC_DB.RAW.SENSOR_READINGS_RAW AS target
            USING (
                SELECT *
                FROM ATMOSYNC_DB.RAW.SENSOR_READINGS_STAGING
                QUALIFY ROW_NUMBER() OVER (
                    PARTITION BY TIMESTAMP, CONTAINER_ID
                    ORDER BY INGESTED_AT
                ) = 1
            ) AS source
            ON target.TIMESTAMP = source.TIMESTAMP
            AND target.CONTAINER_ID = source.CONTAINER_ID

            WHEN NOT MATCHED THEN INSERT (
                TIMESTAMP, CONTAINER_ID, COMMODITY,
                TEMPERATURE, HUMIDITY, VIBRATION,
                TEMPERATURE_STATUS, HUMIDITY_STATUS,
                CONTAINER_STATUS, INGESTED_AT
            )
            VALUES (
                source.TIMESTAMP, source.CONTAINER_ID,
                source.COMMODITY, source.TEMPERATURE,
                source.HUMIDITY, source.VIBRATION,
                source.TEMPERATURE_STATUS,
                source.HUMIDITY_STATUS,
                source.CONTAINER_STATUS, source.INGESTED_AT
            )
        """)

        # Step 5: Clear staging after successful merge
        cursor.execute("""
            DELETE FROM ATMOSYNC_DB.RAW.SENSOR_READINGS_STAGING
        """)

        # Step 6: Commit Snowflake, then Kafka offsets
        conn.commit()
        consumer.commit()

        print("Snowflake ingestion completed successfully!")
        print("Valid messages processed:", len(valid_records))

    except Exception as e:
        conn.rollback()
        print("Ingestion failed:", e)

    finally:
        cursor.close()
```
